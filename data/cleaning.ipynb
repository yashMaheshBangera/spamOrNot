## Check Pre-training Metric : Class Imbalance

```python
print(f"Number of records: {len(y)}")
print(f"Number of records in each class: \n{y.value_counts()}")
```

<div class="alert alert-block alert-info">
<b> Observation: </b> As can be seen, the data is balanced with 1813 records in class 1 and 2788 records in class 0. This will make the model prone to underfitting on the class 1 and create problems with generalizability. To handle this, we oversample the minority class. 
</div>

## Oversample the Minority Class

```python
from imblearn.over_sampling import RandomOverSampler
oversample = RandomOverSampler(sampling_strategy='minority')
```

```python
X, y = oversample.fit_resample(X, y)
```

```python
print(f"Number of records: {len(y)}")
print(f"Number of records in each class after oversampling: \n{y.value_counts()}")
```

## Data skewness 

```python
X.describe()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
num_columns = 3
num_rows = len(columns) // num_columns
fig, axes = plt.subplots(num_rows, num_columns, figsize=(15, num_rows * 4))
axes = axes.flatten()

for i, col in enumerate(columns):
    sns.histplot(df[col], kde=True, ax=axes[i])
    axes[i].set_title(f'Distribution: {col}')

# 5. Clean up: Remove any empty subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

<div class="alert alert-block alert-info">
<b>Observation: </b> Most of the columns above appear to be skewed. This will definitely make the model not generalize well to unseen data. Since we have most of the data to be right skewed, we can remove skewness using Log Transformation
</div>

```python
import numpy as np

log_col=[]
transformed_df = {}
for i in range(len(columns)):
    transformed_df[f'{columns[i]}_log'] = np.log1p(df[columns[i]])
    log_col.append(f'{columns[i]}_log')
```

```python
df_transformed = pd.DataFrame(transformed_df, columns=log_col, index=df.index)
```

```python
df_transformed
```

```python
num_columns = 3
num_rows = len(log_col) // num_columns
fig, axes = plt.subplots(num_rows, num_columns, figsize=(15, num_rows * 4))
axes = axes.flatten()

for i, col in enumerate(log_col):
    sns.histplot(df_transformed[col], kde=True, ax=axes[i])
    axes[i].set_title(f'Distribution: {col}')

# 5. Clean up: Remove any empty subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
df_transformed
```

<div class="alert alert-block alert-info">
<b> Observation :</b> As per the above plots post the transformation, we still have skewed distributions for most columns. This is due to the high number of 0 values we have in the data.
</div>

## Feature Engineering

```python
df.columns
```

<b>Observation : </b>It is often seen that many spam emails contain multiple punctuations. For example: "Buy this TV at just $40.99 !!!!!". Hence, we can create a new feature that will be helpful for our classification problem. The feature is called `punctuation_count` and it is the addition of the character based columns. 

```python
df['punctuation_count'] = df[['char_freq_;','char_freq_(','char_freq_[','char_freq_!','char_freq_$','char_freq_#']].sum(axis=1)
```
